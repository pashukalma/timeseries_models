#### Time Series Concepts Recap

**Time Series** statistical data, a collection of numerical measurements called observations, $x_1, \dots, x_t, \dots, x_n$ in $\mathbb{R}$, indexed at time $t = 1, \dots, n$.
A probabilistic model describing an event is called a **stochastic process** that is collection of r.v.s, {$X_t$} indexed by time parameter $t$.

Deterministic Dependence, Trend and Seasonality:
*   **Trend** can be linear or nonlinear, and sometimes monotone in time.
*   **Seasonal** is a critical component with periodicity at fixed time interval.
*   $X_t$ at different times may be dependent on each other ($X_{t+1}$ may be dependent or independent of the preceding $X_t$), i.e., stock price forecast.
*   $X_t$ correlation may remain constant or change as time distance between observations increases. For this, use **auto-covariance function**.

The time series analysis helps us understand and explore the deterministic and stochastic dependencies of the stochastic process that generated the data. By developing statistical tools, algorithms, and models, aim to detect the trend and seasonal variations, and describe the correlation structure within the same time between two different time series.

##### **Discrete-Time Systems**

A discrete-time system model processes a discrete-time input signal to produce a discrete-time output signal. Both the input and output signals are sequences, typically indexed by integers representing time steps. A system transforms an input sequence $x[n]$ into an output sequence $y[n]$.

Discrete-Time System Characteristics

In a discrete-time system, the output $y(t)$ is related to the input $u(t)$ by the equation $y(t) = \sum_{k} h(k)u(t-k)$.
This is known as **convolution** between the input signal $u(t)$ and the system's impulse response (kernel) $h(k)$.

##### Linearity and Time Invariance

The linear system defined by convolution with a given kernel $h(k)$ is both **linear** and **time-invariant**.

**Linearity:**
For any input signals $u_1(t)$, $u_2(t)$ and scalars $\alpha_1, \alpha_2 \in \mathbb{R}$, if the output of the system $H$ for input $u$ is $H(u)$, then
- $H(\alpha_1 u_1(t) + \alpha_2 u_2(t)) = \sum_{k} h(k) (\alpha_1 u_1(t-k) + \alpha_2 u_2(t-k)) \\ = \alpha_1 \sum_{k} h(k)u_1(t-k) + \alpha_2 \sum_{k} h(k)u_2(t-k) \\ = \alpha_1 H(u_1(t)) + \alpha_2 H(u_2(t))$

**Time Invariance:**
Let a shifted input be $u'(t) = u(t-T)$ for a given integer shift $T \in \mathbb{Z}$. A time-invariant system's output will also be shifted by the same amount $H(u'(t)) = y(t-T)$

##### **Causality**

The LTI system is **causal** if $h(k) = 0$ for all $k < 0$.
Causality means that the current output $y(t)$ does not depend on future inputs like $u(t+1)$, $u(t+2)$, etc. Causality is a fundamental property of all physically realizable systems. If a causal system also satisfies $h(0) = 0$, we say the system is **strictly causal**, as $y(t)$ can only depend on past inputs in ${u(r) : r < t}$.

##### **Stability**

LTI system is Bounded-Input Bounded-Output (**BIBO**) stable if any bounded input $u(t)$ generates bounded output $y(t)$.
It is show that $\sum |h(t)| < \infty$. If $h(t)$ satisfies the stronger condition $\sum t \cdot |h(t)| < \infty$, refer to it as **strict stability**. This can occur if $h(t)$ is a decaying exponential function, $h(t) = \sum A_j e^{-\lambda_j t}$ for constants $A_j, \lambda_j > 0$.

##### **Transfer Function and q-Transform**

Express the input-output operator $H$ or **convolution kernel** $h(\cdot)$ in terms of the **transfer function** $H(q)$ defined as $H(q) = \sum_{k} h(k)q^{-k}$, where $q \in \mathbb{C}$.
The **q-transform** maps a convolution kernel to a complex-valued function of complex variables.

Consider an LTI system $H$ characterized by a convolution kernel $h(k)$ such that $h(k) = \begin{cases} 1 & \text{for } k \ge 0 \\ 0 & \text{for } k < 0 \end{cases}$.
The q-transform of this kernel is given by
$H(q) = \sum_{k=0}^{\infty} q^{-k} = \frac{1}{1 - q^{-1}}$, where the equality holds for all $|q| > 1$.

If a system has a **rational q-transfer function**, given by $H(q) = N(q)/D(q)$, the system is **stable** if and only if all **poles** of $H(q)$ - that is, the roots of the denominator $D(q)$ - lie outside the **unit disc** in the complex plane.

##### **Frequency Response (Eigen Functions**)

A fundamental property of linear systems is that when the input signal is a trigonometric function, the output will also be a trigonometric function of the same frequency, but with a magnitude and phase shift.

The function $u(t) = e^{-i\omega t}$ is an eigenfunction of any linear system.
Given a linear system with convolution kernel $g(\cdot)$, let the input signal $u(t) = e^{-i\omega t}$, then the output of the system $G$ satisfies
$y(t) = \sum_{k} g(k)u(t-k) = \sum_{k} g(k)e^{-i\omega(t-k)} = e^{-i\omega t}\left(\sum_{k} g(k)e^{-i\omega k}\right) = e^{-i\omega t}G(e^{-i\omega})$

$G(q)$ follows the definition of q-transform, i.e., $G(q) = \sum_{k} g(k)q^{-k}$.
If the input signal is $u(t) = \text{Re}(e^{-i\omega t}) = \cos(\omega t)$, then $y(t) = \text{Re}( e^{-i\omega t}G(e^{-i\omega})) = |G(e^{-i\omega})|\cos(\omega t + \phi)$, where $\phi$ is the phase of $G(e^{-i\omega})$.

Use Euler's formula $e^{ix} = \cos(x) + i\sin(x)$. The output of a trigonometric input signal is a trigonometric signal of the same frequency but with magnitude and phase shift.

##### **Discrete Fourier Transforms**

Given a signal ${u(t)}$ for $t =1,2,\dots,N$, let $U_N(\omega) = \frac{1}{\sqrt{N}} \sum_{t=1}^{N} u(t)e^{-i\omega t}$. In addition, the sequence ${U_N(2\pi/N \cdot k)}$ for $k=1,2,\dots,N$ is the Discrete Fourier Transform of signal ${u(\cdot)}$.

The DFT provides a frequency summary of the signal $u(t)$ at the fundamental frequencies $2\pi/N, 4\pi/N, \dots,2\pi$ and can be thought of as the standard Fourier transform of $u(t)$ if its $N$ samples were repeated such that $u(t)$ was periodic.

The DFT encodes all information in the signal $u(\cdot)$ in the sense that the original signal can be recovered using inverse DFT:
$u(t) = \frac{1}{\sqrt{N}} \sum_{k=1}^{N} U_N(2\pi/N \cdot k) e^{i \cdot 2\pi/N \cdot k \cdot t}$

DFT has several useful properties:

- First $|U_N(2\pi/N \cdot k)|^2$, $k =1,\dots,N$ encodes the energy of the signal at each fundamental frequency $2\pi/N \cdot k$. By plotting $|U_N(2\pi/N \cdot k)|^2$ for each $k$, obtain the visualization for the frequency content of the signal $u(\cdot)$, known as **periodogram**. The first property of the DFT is that the energy content of the DFT and the original signal are equivalent, which is given by **Parseval's Theorem**:
For a signal ${u(t)}$ with DFT ${U_N(2\pi/N \cdot k)}$, $k =1,\dots,N$, this holds
$\sum_{t=1}^{N} |u(t)|^2 = \sum_{k=1}^{N} |U_N(2\pi/N \cdot k)|^2$.

- The second property is that DFT obeys periodicity, i.e.,
$U_N(\omega +2\pi) = U_N(\omega)$
and if $u(\cdot)$ is a real signal, then it satisfies $U_N(\omega) = U_N(\omega)^\ast$

##### **DFT and Filters**

Suppose $|u(t)| \le C_U$ for all $t$ and $\sum_{t} t \cdot |h(t)| \le C_H < \infty$ for a causal linear system. Then the DFT of the output signal satisfies that
$Y_N(\omega) = H(e^{i\omega}) \cdot U_N(\omega) + R_N(\omega)$, where the **residue term** $|R_N(\omega)| \le C_U \cdot C_N / \sqrt{N}$.

##### **DFT: Cosine Signal**

Let $u(t) = A\cos(\omega_0 t)$ for $t =1,\dots,N$, where $\omega_0 = 2\pi/N_0$. Then for a given integer $s$, consider DFT for the signal $u(\cdot)$ with $N = s \cdot N_0$, which gives
- $U_N(\omega) = \frac{1}{\sqrt{N}} \sum_{t=1}^{N} A\cos(\omega_0 t) e^{-i\omega t}$.

This example explores the properties of the DFT when the signal is a perfect cosine wave sampled over an integer number of its periods.

##### **DFT: Perfectly Periodic Signal**

Suppose $u(t)$ has a period of $N_0$, i.e., $u(t) = u(t + N_0)$ for all $t$. First consider a truncation of this signal with length $N_0$ defined by
$u_{\text{tilda}}(t) = u(t), \quad \text{for } t =1, \dots, N_0$.

The **DFT** of $u_{\text{tilda}}(t)$ is $A_r := U_{\text{tilda}, N_0}(2\pi/N_0 \cdot r) = \frac{1}{\sqrt{N_0}} \sum_{t=1}^{N_0} u_{\text{tilda}}(t) e^{-i \cdot 2\pi/N_0 \cdot r \cdot t}$ for $r = 1,\dots,N_0$.

Analyze and reconstruct $u_{\text{tilda}}(t)$ by the inverse of the **DFT**.

This example aims to show that the frequency component $U_N(\omega)$ is non-zero only for $\omega = 2\pi/N_0 \cdot r$, where $r =1,\dots,N_0$ (i.e., at the fundamental frequency and its harmonics).

##### **Linearity**

A discrete-time linear system $T$ satisfies two properties:

1.  Additivity, the response to a sum of inputs is the sum of the responses to each input individually. $T\{x_1[n] + x_2[n]\} = T\{x_1[n]\} + T\{x_2[n]\}$.

2.  Homogeneity (Scaling), the response to a scaled input is the scaled response of the input. $T\{a \cdot x[n]\} = a \cdot T\{x[n]\}$.

Combining these, for any inputs $x_1[n]$, $x_2[n]$ and scalars $a$, $b$
$T\{a \cdot x_1[n] + b \cdot x_2[n]\} = a \cdot T\{x_1[n]\} + b \cdot T\{x_2[n]\}$

```python
# Define a simple linear discrete-time system: y[n] = 2 * x[n]
def linear_system(x):
    return 2 * x

# Input signals
x1, x2 = np.array([1, 2, 3]), np.array([4, 5, 6])
# Scalars
a, b = 3, 0.5

# Test Additivity:
y1, y2 = linear_system(x1), linear_system(x2)
y_sum_individual = y1 + y2

y_sum_input = linear_system(x1 + x2)

# Test Homogeneity:
y_scaled_individual = a * linear_system(x1)
y_scaled_input = linear_system(a * x1)
```

##### **Time Invariance**

A discrete-time system time-invariant $T$ is a time shift in the input signal causing an identical time shift in the output signal.
If $T\{x[n]\} = y[n]$, then $T\{x[n - k]\} = y[n - k]$ for any integer shift $k$.
The system's behavior does not change over time. The same input at a different time produces the same output, just shifted in time.

```python
# Define a simple time-invariant system: y[n] = x[n] + x[n-1]
def time_invariant_system(x_input, delay=0):
    # Pad the input with zeros for the delayed samples
    x_padded = np.pad(x_input, (delay, 0), 'constant', constant_values=0)
    y = np.zeros_like(x_input, dtype=float)
    for n in range(len(x_input)):
        if n - delay >= 0:
            y[n] = x_padded[n] + x_padded[n-1+delay] # x[n] + x[n-1] logic
        else:
            y[n] = x_padded[n]
            # handle initial conditions if x[n-1] is out of bounds
    return y

# Input signal
x = np.array([1, 2, 3, 4, 5])

# Original output
y_original = time_invariant_system(x)

# Shift the input by k = 2 samples
k = 2
x_shifted = np.array([0, 0, 1, 2, 3]) # x[n-2]
y_shifted_input = time_invariant_system(x_shifted)

# Expected output (original output shifted by k)
y_expected_shifted = np.pad(y_original, (k, 0), 'constant', constant_values=0)[ : len(y_original) + k]

# Compare y_shifted_input with a manually shifted y_original
# We consider the length of the outputs for comparison due to padding
# For true comparison, we ensure the system handles boundaries consistently
# Re-run for a more direct comparison by checking y[n-k] where n-k is valid
# Consider a segment where the shift is fully applied within the input bounds

# Define a system where output for index `n` depends on `x[n]` and `x[n-1]`
# For illustration, let's simplify to y[n] = x[n] + x[n-1]
# This assumes x[-1]=0, etc.

def system_kernel(x_val, x_prev_val):
    return x_val + x_prev_val

# Compute original output
y_orig_val = [system_kernel(x[0], 0)] # Assuming x[-1]=0
for i in range(1, len(x)):
    y_orig_val.append(system_kernel(x[i], x[i-1]))
y_original_simpler = np.array(y_orig_val)

# Compute shifted input ; x[n-k]
x_shifted_simpler = np.pad(x, (k, 0), 'constant', constant_values=0)[:len(x)]

# Compute output of shifted input
y_from_shifted_input_val = [system_kernel(x_shifted_simpler[0], 0)]
for i in range(1, len(x_shifted_simpler)):
    y_from_shifted_input_val.append(system_kernel(x_shifted_simpler[i], x_shifted_simpler[i-1]))
y_from_shifted_input_simpler = np.array(y_from_shifted_input_val)

# Compute expected output (original output shifted)
y_expected_shifted_simpler = np.pad(y_original_simpler, (k, 0), 'constant', constant_values=0)[:len(y_original_simpler)]

# Compare relevant parts (ignoring initial zeros from padding)
# We compare the non-zero parts of the outputs

# Check if the outputs match after the shift has taken effect
# For n >= k, we expect T{x[n-k]} to be y[n-k]
# So, y_from_shifted_input_simpler[k:] should match y_original_simpler[:-k] (if shifted left)
# Or, y_from_shifted_input_simpler should match y_original_simpler shifted right

# Let compare y_from_shifted_input_simpler with y_original_simpler shifted by k
# This means comparing y_from_shifted_input_simpler[i] with y_original_simpler[i-k] for i >= k

comparison_result = np.allclose(y_from_shifted_input_simpler[k:], y_original_simpler[:-k])
```

##### Visualizing Causality

```python
# Input signal (e.g., a pulse)
x_causality = np.zeros(20)
x_causality[5:10] = 1 # A pulse from n=5 to n=9

# Define a causal system: y[n] = x[n] + 0.5 * x[n-1]
def causal_system(x_input):
    y = np.zeros_like(x_input, dtype=float)
    for n in range(len(x_input)):
        y[n] = x_input[n]
        if n-1 >= 0:
            y[n] += 0.5 * x_input[n-1]
    return y

# Define a non-causal system: y[n] = x[n+1] + x[n]
def non_causal_system(x_input):
    y = np.zeros_like(x_input, dtype=float)
    for n in range(len(x_input)):
        y[n] = x_input[n]
        if n+1 < len(x_input):
            y[n] += x_input[n+1]
        # For the last sample, x[n+1] is assumed to be 0 or handled by boundary
    return y

y_causal = causal_system(x_causality)
y_non_causal = non_causal_system(x_causality)
```

##### Visualizing Stability (BIBO Stability)

```python
# Input signal: Unit Impulse (bounded input)
impulse = np.zeros(50)
impulse[0] = 1 # Impulse at n=0

# Define a stable system (e.g., y[n] = 0.8 * y[n-1] + x[n])
def stable_system(x_input):
    y = np.zeros_like(x_input, dtype=float)
    for n in range(len(x_input)):
        if n == 0:
            y[n] = x_input[n]
        else:
            y[n] = 0.8 * y[n-1] + x_input[n]
    return y

# Define an unstable system (e.g., y[n] = 1.2 * y[n-1] + x[n])
def unstable_system(x_input):
    y = np.zeros_like(x_input, dtype=float)
    for n in range(len(x_input)):
        if n == 0:
            y[n] = x_input[n]
        else:
            y[n] = 1.2 * y[n-1] + x_input[n]
    return y

y_stable = stable_system(impulse)
y_unstable = unstable_system(impulse)
```

#### Transfer Function

For a **LTI** discrete-time system, the **transfer function**, denoted as $H(z)$, is the ratio of the Z-transform of the output signal $Y(z)$ to the Z-transform of the input signal $X(z)$, assuming zero initial conditions, often represented in the z-domain as $H(z) = \frac{Y(z)}{X(z)}$.

The transfer function can also be derived from the system's difference equation. If an LTI system is described by the difference equation
$y[n] + a_1 y[n-1] + \dots + a_N y[n-N] = b_0 x[n] + b_1 x[n-1] + \dots + b_M x[n-M]$, its transfer function $H(z)$ is given by
$H(z) = \frac{B(z)}{A(z)} = \frac{b_0 + b_1 z^{-1} + \dots + b_M z^{-M}}{1 + a_1 z^{-1} + \dots + a_N z^{-N}}$.

The **Z-transform** is a tool that converts a discrete-time signal from the time domain to the complex frequency domain, z-domain. It is the discrete-time equivalent of the Laplace transform used for continuous-time signals.

$X(z) = \sum_{n=0}^{\infty} x[n] z^{-n}$,

#### Visualization of DFT

The DFT converts a finite sequence of discrete-time samples from the time domain to the frequency domain. It decomposes the sequence into its constituent frequencies, revealing the frequency components present in the signal.
For a discrete-time sequence $x[n]$ of length $N$ (where $n = 0, 1, \dots, N-1$), is given by
$X[k] = \sum_{n=0}^{N-1} x[n] e^{-j 2\pi k n / N}$,
where $k = 0, 1, \dots, N-1$ represents the frequency index.
The Inverse DFT reconstructs the original time-domain sequence from its DFT coefficients
$x[n] = \frac{1}{N} \sum_{k=0}^{N-1} X[k] e^{j 2\pi k n / N}$.
The DFT is widely used in signal processing for spectrum analysis, filtering, and compression. The Fast Fourier Transform is an efficient algorithm for computing the DFT.

```python
# Generate a sample signal (e.g., sum of two sinusoids)
fs = 100 # Sampling frequency
t = np.arange(0, 1, 1/fs) # Time vector

f1, f2 = 5, 20 # Frequency 1, Frequency 2
x_signal = 2 * np.sin(2 * np.pi * f1 * t) + 1 * np.sin(2 * np.pi * f2 * t) + 0.5 * np.random.randn(len(t)) # Add some noise

N = len(x_signal) # Length of the signal

# Compute the Discrete Fourier Transform (DFT) using numpy's FFT
X_k = np.fft.fft(x_signal)

# Compute the corresponding frequencies
freq = np.fft.fftfreq(N, 1/fs)

# Only take the positive frequencies and scale the magnitude
positive_freq_indices = np.where(freq >= 0)
freq_positive = freq[positive_freq_indices]
magnitude_spectrum = 2/N * np.abs(X_k[positive_freq_indices]) # Multiply by 2/N for amplitude scaling

# Plot the time-domain signal
plt.figure(figsize=(12, 8), dpi=60) ; plt.subplot(2, 1, 1)
plt.plot(t, x_signal)
plt.title('Time Domain Signal') ; plt.xlabel('Time (s)')
plt.ylabel('Amplitude') ; plt.grid(True, linestyle='--', alpha=0.7)

# Plot the magnitude spectrum
plt.subplot(2, 1, 2)
plt.stem(freq_positive, magnitude_spectrum, basefmt=' ')
plt.title('Magnitude Spectrum (DFT)')
plt.xlabel('Frequency (Hz)') ; plt.ylabel('Magnitude')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xlim(0, fs/2) # Show frequencies up to Nyquist frequency
plt.suptitle('Discrete Fourier Transform Example', y=1.02, fontsize=16)
plt.tight_layout() ; plt.show()
```

**Periodogram** as an estimate of the spectral density of a signal, is essentially the magnitude squared of the DFT of the signal, normalized by the number of samples.
The periodogram $P(k)$ is defined as $P(k) = \frac{1}{N} |X[k]|^2$, or, more commonly, as the power spectral density (PSD)
$P_{xx}(f) = \frac{1}{N} |DFT(x[n])|^2 = \frac{1}{N} |X[k]|^2$,
where $N$ is the number of samples and $X[k]$ is the DFT of $x[n]$. The periodogram is a simple and common method for estimating the power spectrum of a signal, showing how the power of a signal is distributed over different frequencies.

```python
# Reuse the sample signal from the DFT example
fs = 100 # Sampling frequency
t = np.arange(0, 1, 1/fs) # Time vector

f1 = 5  # Frequency 1
f2 = 20 # Frequency 2
x_signal = 2 * np.sin(2 * np.pi * f1 * t) + 1 * np.sin(2 * np.pi * f2 * t) + 0.5 * np.random.randn(len(t)) # Add some noise

# Compute the periodogram
frequencies, power_spectral_density = periodogram(x_signal, fs=fs)

# Plot the periodogram
plt.figure(figsize=(12, 5), dpi=60)
plt.semilogy(frequencies, power_spectral_density)
plt.title('Periodogram (Power Spectral Density)')
plt.xlabel('Frequency (Hz)') ; plt.ylabel('Power/Frequency (dB/Hz)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xlim(0, fs/2) # Show frequencies up to Nyquist frequency
plt.suptitle('Periodogram Example', y=1.02, fontsize=16)
plt.tight_layout() ; plt.show()
```

The DFT possesses several important properties that are essential for its application in signal processing:

1.  **Linearity:** If $x_1[n] \xrightarrow{\text{DFT}} X_1[k]$ and $x_2[n] \xrightarrow{\text{DFT}} X_2[k]$, then $a x_1[n] + b x_2[n] \xrightarrow{\text{DFT}} a X_1[k] + b X_2[k]$.

2.  **Periodicity:** Both the time-domain sequence $x[n]$ and its DFT $X[k]$ are periodic with period $N$ (the length of the sequence). That is, $x[n+N] = x[n]$ and $X[k+N] = X[k]$.

3.  **Symmetry Properties:** For a real-valued input signal $x[n]$, its DFT $X[k]$ exhibits Hermitian symmetry $X[k] = X^*[N-k]$ (where $^*$ denotes complex conjugate). If $x[n]$ is real and even, $X[k]$ is real and even. If $x[n]$ is real and odd, $X[k]$ is imaginary and odd.

4.  **Time Shift:** If $x[n] \xrightarrow{\text{DFT}} X[k]$, then $x[n-m] \xrightarrow{\text{DFT}} X[k] e^{-j 2\pi k m / N}$. A circular shift in time corresponds to a linear phase shift in the frequency domain.

5.  **Frequency Shift:** If $x[n] \xrightarrow{\text{DFT}} X[k]$, then $x[n] e^{j 2\pi n m / N} \xrightarrow{\text{DFT}} X[k-m]$. Multiplication by a complex exponential in the time domain corresponds to a circular shift in the frequency domain.

6.  **Convolution Property (Circular Convolution):** Circular convolution in the time domain corresponds to multiplication in the frequency domain. If $x[n] \circledast h[n] \xrightarrow{\text{DFT}} X[k] H[k]$, where $\circledast$ denotes circular convolution.

7.  **Multiplication Property (Circular Convolution in Time):** Multiplication in the time domain corresponds to circular convolution in the frequency domain. If $x_1[n] x_2[n] \xrightarrow{\text{DFT}} \frac{1}{N} (X_1[k] \circledast X_2[k])$.

8.  **Parseval's Theorem:** This theorem relates the energy of a signal in the time domain to its energy in the frequency domain. $\sum_{n=0}^{N-1} |x[n]|^2 = \frac{1}{N} \sum_{k=0}^{N-1} |X[k]|^2$.

Demonstrating Periodicity of the DFT

The DFT $X[k]$ of a sequence $x[n]$ of length $N$ is periodic with period $N$. That is, $X[k] = X[k+N]$.

```python
# Define a short discrete-time signal
x_periodicity = np.array([1, 2, 3, 4])
N_periodicity = len(x_periodicity) # Length of the signal

# Compute the DFT of the signal
X_k_periodicity = np.fft.fft(x_periodicity)

# Create an extended range for k to show periodicity
k_range = np.arange(N_periodicity + 5) # Go beyond N to show periodicity

# Calculate the periodic extension of X[k]
X_k_extended = np.array([X_k_periodicity[k % N_periodicity] for k in k_range])

# Plot the original and extended DFT to visualize periodicity
plt.figure(figsize=(10, 6), dpi=60)
plt.stem(np.arange(N_periodicity), X_k_periodicity.real, linefmt='b-', markerfmt='bo', basefmt=' ', label='X[k] (Original DFT)')
plt.stem(k_range, X_k_extended.real, linefmt='r:', markerfmt='rx', basefmt=' ',
         label='X[k] Extended (Real Part)')
plt.title('Periodicity of the DFT (Real Part)')
plt.xlabel('Frequency Index k') ; plt.ylabel('Magnitude (Real Part)')
plt.xticks(np.arange(0, N_periodicity + 5, 1))
plt.axvline(N_periodicity, color='gray', linestyle='--', label=f'N = {N_periodicity}')
plt.legend() ; plt.grid(True, linestyle='--', alpha=0.7) ; plt.show()

# Explicitly compare values to confirm periodicity
print(f"Original DFT X[k]: {X_k_periodicity}")
print(f"X[k] for k=0: {X_k_periodicity[0]:.4f}")
print(f"X[k] for k={N_periodicity} (X[0+N]): {X_k_extended[N_periodicity]:.4f}")
print(f"X[k] for k=1: {X_k_periodicity[1]:.4f}")
print(f"X[k] for k={N_periodicity+1} (X[1+N]): {X_k_extended[N_periodicity+1]:.4f}")

# Check if X[0] == X[N], X[1] == X[N+1], etc.
periodicity_check = np.allclose(X_k_periodicity, X_k_extended[:N_periodicity])
print(f"DFT is periodic with period N: {periodicity_check}")
```

DFT Example: Cosine Signal $u(n) = A cos(w_0 n)$

For a signal $u[n] = A \cos(\omega_0 n)$ with a fundamental period $N_0$ samples, such that $\omega_0 = 2\pi / N_0$ (normalized angular frequency),
and we take $N$ samples where $N = s \cdot N_0$ for some integer $s$ (meaning we capture an exact integer number of cycles).
The DFT $X[k]$ of such a signal will typically have two non-zero components (for real $A$)
*   At $k = s$
*   At $k = N - s$ (which is equivalent to $-s$ due to DFT periodicity)

```python
# Define Signal Parameters
A = 1.0       # Amplitude
N0 = 10       # Fundamental period in samples (e.g., 10 samples per cycle)
s = 3         # Number of full cycles, N = s * N0

w0 = 2 * np.pi / N0 # Normalized angular frequency
N = s * N0    # Total number of samples for the DFT

# Generate the Cosine Signal
n = np.arange(N) # Discrete time indices from 0 to N-1
u_n = A * np.cos(w0 * n)

# Compute the DFT
X_k = np.fft.fft(u_n)

# Compute Corresponding Frequencies
# The DFT frequencies are k = 0, 1, ..., N-1
# The corresponding physical frequencies (if fs=1) would be k/N

# Plotting
plt.figure(figsize=(12, 8), dpi=60)

# Plot Time-Domain Signal
plt.subplot(2, 1, 1) ; plt.stem(n, u_n, basefmt=' ')
plt.title(f'Time-Domain Signal: u[n] = {A} cos( (2pi/{N0}) n ) for N={N}')
plt.xlabel('n (samples)') ; plt.ylabel('Amplitude')
plt.grid(True, linestyle='--', alpha=0.7)

# Plot Magnitude Spectrum (DFT)
plt.subplot(2, 1, 2)
k_values = np.arange(N) # DFT index k
magnitude_spectrum = np.abs(X_k)

plt.stem(k_values, magnitude_spectrum, basefmt=' ')
plt.title('Magnitude Spectrum (DFT)')
plt.xlabel('Frequency Index k') ; plt.ylabel('|X[k]|')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xlim(-1, N) # Show all k values
plt.xticks(np.arange(0, N, N0/s))

plt.suptitle('DFT of a Cosine Signal Example', y=1.02, fontsize=16)
plt.tight_layout() ; plt.show()

print(f"Signal parameters: A={A}, N0={N0}, s={s}, N={N}, w0={w0:.4f}")
print(f"DFT X[k] for k={s}: {X_k[s]:.4f}")
print(f"DFT X[k] for k={N-s}: {X_k[N-s]:.4f}")
print(f"Expected peak magnitude: A * N / 2 = {A * N / 2}")
```

DFT Example: Perfectly Periodic Signal

Consider a discrete-time signal $u[n]$ that is periodic with a fundamental period $N_0$, meaning $u[n] = u[n+N_0]$ for all $n$.
If we compute the DFT over an observation window of length $N$, where $N$ is an integer multiple of $N_0$ (i.e., $N = s \cdot N_0$ for some integer $s$), then the DFT $X[k]$ will ideally only have non-zero components at the fundamental frequency and its harmonics. All other frequency components will be zero, assuming no noise.

```python
# Define Signal Parameters
N0 = 16       # Fundamental period of the signal in samples
s = 4         # Number of full periods in the observation window
N = s * N0    # Total number of samples for the DFT (e.g., 64 samples)

# Generate a periodic signal (e.g., a square wave or sum of harmonics)
# For simplicity, let's create a signal that is a sum of its fundamental and a third harmonic
n = np.arange(N) # Discrete time indices from 0 to N-1

# Fundamental frequency component
f_fundamental = 1 / N0 # cycles per sample
u_fundamental = 1.0 * np.sin(2 * np.pi * f_fundamental * n)

# Third harmonic component
f_harmonic = 3 / N0 # 3rd harmonic
u_harmonic = 0.5 * np.sin(2 * np.pi * f_harmonic * n)

u_n_periodic = u_fundamental + u_harmonic # Our perfectly periodic signal

# Compute the DFT
X_k_periodic = np.fft.fft(u_n_periodic)

# Compute Corresponding Frequencies (for plotting)
k_values = np.arange(N) # DFT index k

# Plotting
plt.figure(figsize=(12, 8), dpi=60)

# Plot Time-Domain Signal
plt.subplot(2, 1, 1) ; plt.plot(n, u_n_periodic)
plt.title(f'Time-Domain Signal: Perfectly Periodic (N0={N0}, N={N})')
plt.xlabel('n (samples)') ; plt.ylabel('Amplitude')
plt.grid(True, linestyle='--', alpha=0.7) ; plt.xlim(0, N-1)

# Plot Magnitude Spectrum (DFT)
plt.subplot(2, 1, 2)
magnitude_spectrum_periodic = np.abs(X_k_periodic)

plt.stem(k_values, magnitude_spectrum_periodic, basefmt=' ')
plt.title('Magnitude Spectrum (DFT) of Periodic Signal')
plt.xlabel('Frequency Index k') ; plt.ylabel('|X[k]|')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xlim(-1, N) # Show all k values

# Highlight expected peaks (fundamental and 3rd harmonic)
expected_k1 = s * 1 # k for fundamental frequency
expected_k2 = s * 3 # k for 3rd harmonic

plt.axvline(x=expected_k1, color='red', linestyle=':', label=f'Fundamental (k={expected_k1})')
plt.axvline(x=N - expected_k1, color='red', linestyle=':')
plt.axvline(x=expected_k2, color='green', linestyle=':', label=f'3rd Harmonic (k={expected_k2})')
plt.axvline(x=N - expected_k2, color='green', linestyle=':')
plt.suptitle('DFT of a Perfectly Periodic Signal Example', y=1.02, fontsize=16)
plt.legend() ; plt.tight_layout() ; plt.show()

print(f"Signal parameters: N0={N0}, s={s}, N={N}")
print(f"DFT X[k] for fundamental (k={expected_k1}): {X_k_periodic[expected_k1]:.4f}")
print(f"DFT X[k] for fundamental conjugate (k={N-expected_k1}): {X_k_periodic[N-expected_k1]:.4f}")
print(f"DFT X[k] for 3rd harmonic (k={expected_k2}): {X_k_periodic[expected_k2]:.4f}")
print(f"DFT X[k] for 3rd harmonic conjugate (k={N-expected_k2}): {X_k_periodic[N-expected_k2]:.4f}")
```

DFT Example: Sinusoidal Signal with Spectral Leakage

This example demonstrates what happens when the length of the DFT window $N$ is not an integer multiple of the signal's period $N_0$. In such cases, the energy of a single frequency component 'leaks' into adjacent frequency bins, a phenomenon known as **spectral leakage**.

```python
import numpy as np
import matplotlib.pyplot as plt

# Define Signal Parameters
A_leak = 1.0       # Amplitude
f_signal = 10.5    # Signal frequency in Hz
fs_leak = 100      # Sampling frequency in Hz

# Number of samples for the DFT - NOT an exact multiple of the period
# Period T = 1/f_signal. Samples per period N_samples_per_period = fs_leak / f_signal = 100 / 10.5 = 9.52...
# Let N be 100, which is not a multiple of 9.52...
N_leak = 100

# Generate the Sinusoidal Signal
t_leak = np.arange(N_leak) / fs_leak # Discrete time indices
u_n_leak = A_leak * np.sin(2 * np.pi * f_signal * t_leak)

# Compute the DFT
X_k_leak = np.fft.fft(u_n_leak)

# Compute Corresponding Frequencies
freq_bins = np.fft.fftfreq(N_leak, 1/fs_leak)

# Get positive frequencies and scale magnitude
positive_freq_indices_leak = np.where(freq_bins >= 0)
freq_positive_leak = freq_bins[positive_freq_indices_leak]
magnitude_spectrum_leak = 2/N_leak * np.abs(X_k_leak[positive_freq_indices_leak]) # Scale for amplitude

# Plotting
plt.figure(figsize=(12, 8), dpi=60)

# Plot Time-Domain Signal
plt.subplot(2, 1, 1) ; plt.plot(t_leak, u_n_leak)
plt.title(f'Time-Domain Signal: A sin({f_signal} Hz) with N={N_leak} samples')
plt.xlabel('Time (s)') ; plt.ylabel('Amplitude')
plt.grid(True, linestyle='--', alpha=0.7)

# Plot Magnitude Spectrum (DFT)
plt.subplot(2, 1, 2)
plt.stem(freq_positive_leak, magnitude_spectrum_leak, basefmt=' ')
plt.title('Magnitude Spectrum (DFT) with Spectral Leakage')
plt.xlabel('Frequency (Hz)') ; plt.ylabel('|X[k]| (Amplitude)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xlim(0, fs_leak/2) # Show frequencies up to Nyquist frequency
plt.suptitle('DFT of a Sinusoidal Signal with Spectral Leakage Example', y=1.02, fontsize=16)
plt.tight_layout() ; plt.show()

print(f"Signal frequency: {f_signal} Hz")
print(f"Sampling frequency: {fs_leak} Hz")
print(f"Number of samples: {N_leak}")
print(f"Expected peak frequency bin (f_signal * N_leak / fs_leak): {f_signal * N_leak / fs_leak:.2f} (not an integer)")

# Find the peak frequency bin and its neighbors
peak_idx = np.argmax(magnitude_spectrum_leak)
print(f"Actual peak DFT frequency bin: {freq_positive_leak[peak_idx]:.2f} Hz")
print(f"Magnitude at peak: {magnitude_spectrum_leak[peak_idx]:.4f}")
print("Notice how the energy is spread across multiple frequency bins around 10.5 Hz, instead of a single sharp peak.")
```
